Normally, two classical bits are independent.

bit A = 0, bit B = 1

When qubits become entangled, they stop having independent states.
Instead, the ENTIRE SYSTEM has one shared quantum state.

The Bell state is the simplest entangled state.
∣Φ+⟩ = ∣00⟩ + ∣11⟩ / sq.rt (2)​

Meaning the system is:
50% both qubits are 0 & 50% both qubits are 1 (in a SP).

NOTE: we can have other entangled states of diferring probablities for 0 & 1.

An important consequence is that when we measure;
If q0 becomes 0, THEN q1 MUST also become 0 (same for 1).

∣01⟩ & ∣10⟩ never appear so their probabilities are literally 0%.

This is because the Bell State only contains ∣00⟩ & ∣11⟩.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```

```python
qc = QuantumCircuit(2, 2)
```

```python
qc.draw('mpl')
```

```python
# H∣0⟩ = ∣0⟩ + ∣1⟩ / sq.rt(2)

qc.h(0) # Now, the system is H∣0⟩ = ∣00⟩ + ∣10⟩ / sq.rt(2)
```

```python
qc.draw('mpl') # in one branch: q0=0, q1=0
               # in the other branch: q0=1, q1=0
```

```python
qc.cx(0, 1) # Flips q1 if q0 is 1 --> Now, the system is ∣00⟩ + ∣11⟩ / sq.rt(2)

qc.draw("mpl")
```
